```python
import pandas as pd
```

```python
df=pd.read_csv("spam.csv")
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Category'].map({'spam':1,'ham':0})
```

```python
X=df.Message
y=df.Category
```

<h4>Train Test Split</h4>

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

<h4>Text Preprocessing</h4>

```python
from sklearn.feature_extraction.text import TfidfVectorizer
v=TfidfVectorizer()
X_train_count= v.fit_transform(X_train)
```

<h4>Naive Bayes </h4>

```python
from sklearn.naive_bayes import MultinomialNB
model= MultinomialNB()
model.fit(X_train_count,y_train)
```

<h4>Testing An Email</h4>

```python
email=["Hey Raj,Come Lets Playfootball",
       "Congratulations! You have won ₹10,000. Claim your prize now. Call 9876543210. Free offer expires today."
      ]
test=v.transform(email)
```

```python
model.predict(test)
```

<h4>Use Pipelining </h4>


```python
#Combining Multiple steps into single package,we can call using pipeling.fit()
```

```python
from sklearn.pipeline import Pipeline
```

```python
cmb=Pipeline([('Vectorizer',TfidfVectorizer()),
              ('nb',MultinomialNB())
             ])
```

```python
cmb.fit(X_train,y_train)
```

```python
cmb.predict(X_train)
```

```python
cmb.score(X_train,y_train)
```

```python
import joblib
```

```python
joblib.dump(cmb,"PipelineModel")
```
